```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from utils import get_db_engine
```

```python
def load_and_preprocess():
    engine = get_db_engine()
    df = pd.read_sql("SELECT * FROM loan_records", con=engine)

    if 'LoanID' in df.columns:
        df = df.drop(columns=['LoanID'])

    cat_cols = ['Education', 'EmploymentType', 'MaritalStatus', 'HasMortgage', 'HasDependents', 'LoanPurpose', 'HasCoSigner']

    le = LabelEncoder()
    for col in cat_cols:
        df[col] = le.fit_transform(df[col])

    scaler = StandardScaler()

    features = df.drop(columns=['Default'])
    target = df['Default']

    scaled_features = scaler.fit_transform(features)

    clean_df = pd.DataFrame(scaled_features, columns=features.columns)
    clean_df['Default'] = target.values

    return clean_df
```

```python
if __name__ == "__main__":
    df_clean = load_and_preprocess()
    print("✅ Preprocessing Success!")
    print(df_clean.head())
```
