```python
import pandas as pd
from utils import get_db_engine
```

```python
def migrate_data():
    df = pd.read_csv("../data/loan_data.csv")
    engine = get_db_engine()

    df.to_sql('loan_records', con=engine, if_exists='replace', index=False)
    print("🔥 Migration Success!")
```

```python
if __name__ == "__main__":
    migrate_data()
```
